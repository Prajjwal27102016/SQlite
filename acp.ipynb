```python
import pandas as pd
import sqlite3
database = 'basketball.sqlite'
conn = sqlite3.connect(database)
query = pd.read_sql('''SELECT p.id, p.full_name AS player_name, pa.TEAM_NAME, pa.PTS, pa.AST, pa.REB
                        FROM Player p
                        INNER JOIN Player_Attributes pa
                        ON p.id = pa.ID
                        LIMIT 20
                        ''',conn)
print(query)
```
